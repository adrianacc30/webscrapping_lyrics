```python
import lyricsgenius
import time
import re
import os
from getpass import getpass

# Read the token from your environmentGENIUS_TOKEN = os.environ.get("GENIUS_TOKEN") or getpass("Genius access token (input hidden): ")
if not GENIUS_TOKEN.strip():
    raise ValueError("A Genius API access token is required.")

# Initialize Genius client
genius = lyricsgenius.Genius(GENIUS_TOKEN)

# Optional: turn off verbose output and auto‑removing of featured artists
genius.verbose = False
genius.remove_section_headers = True  # Removes [Chorus], [Verse] labels

# Song titles to search for
song_titles = [
    "pixelated kisses",
    "sojourn",
    "ew",
    "sanctuary",
    "love you less",
    "glimpse of us",
    "if it only gets better",
    "past wont leave my bed",
    "hotel california",
    "daylight",
    "yeah right",
    "worldstar money",
    "will he",
    "horses to water",
    "tarmac",
    "tick tock",
    "pretty boy",
    "plastic taste",
    "afterthought",
    "attention",
    "blah blah",
    "like you do",
    "reanimator",
    "night rider",
    "last of a dying breed",
    "gimme love",
    "die for you",
    "demons",
    "1am freestyle",
    "slow dancing in the dark"
]

artist = "Joji"

# Store results
lyrics_dict = {}

print(f"🎵 Fetching lyrics for {len(song_titles)} songs by {artist}...\n")

for idx, title in enumerate(song_titles, 1):
    print(f"[{idx}/{len(song_titles)}] Searching: {title}")
    try:
        # Search for the song
        song = genius.search_song(title, artist)
        if song:
            lyrics = song.lyrics
            # Optional: clean up extra whitespace
            lyrics = re.sub(r'\n\s*\n', '\n', lyrics).strip()
            lyrics_dict[title] = lyrics
            print(f"  ✅ Found! ({len(lyrics.splitlines())} lines)")
        else:
            lyrics_dict[title] = "NOT FOUND"
            print(f"  ❌ Not found.")
    except Exception as e:
        error_name = type(e).__name__
        lyrics_dict[title] = f"ERROR: {error_name}"
        print(f"  ⚠️ Error: {error_name}")
    
    # Be polite – wait 1 second between requests
    time.sleep(1)

# Save all lyrics to a single text file
output_file = "joji_lyrics.txt"
with open(output_file, "w", encoding="utf-8") as f:
    for title, lyrics in lyrics_dict.items():
        f.write(f"\n--- {title.upper()} ---\n")
        f.write(lyrics)
        f.write("\n\n")

print(f"\n✅ Done! All lyrics saved to '{output_file}'")
```
